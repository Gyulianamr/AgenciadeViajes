```python
python preprocesamiento.py
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import string

# Descargar recursos necesarios de NLTK
nltk.download('punkt')
nltk.download('stopwords')

def preprocess_text(text):
    """
    Función para preprocesar un texto:
    1. Tokeniza el texto.
    2. Elimina stopwords y puntuación.
    3. Convierte todo a minúsculas.
    """
    # Convertir a minúsculas
    text = text.lower()

    # Tokenizar el texto
    tokens = word_tokenize(text)

    # Definir stopwords en español
    stop_words = set(stopwords.words('spanish'))

    # Filtrar stopwords y puntuación
    filtered_tokens = [
        word for word in tokens
        if word not in stop_words and word not in string.punctuation
    ]

    return filtered_tokens

# Ejemplo de uso
if __name__ == "__main__":
    # Texto de ejemplo
    raw_texts = [
        "El cirujano robótico realizó una operación compleja con precisión milimétrica.",
        "La tecnología avanzada permite mejorar resultados en pacientes."
    ]

    # Preprocesar todos los textos
    preprocessed_texts = [preprocess_text(text) for text in raw_texts]

    # Mostrar resultados
    for i, processed_text in enumerate(preprocessed_texts, 1):
        print(f"Texto {i} preprocesado: {processed_text}")
```
